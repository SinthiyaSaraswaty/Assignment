```python
reg.fit(df[['Weight']], df[['Price']])
```

```python
df2 = df.copy()
df2.head()
```

```python
df2['Predicted_Price']= reg.predict(df[['Weight']])
df2.head()
```

```python
df2['residuals'] = df2['Price'] - df2['Predicted_Price']
df2.head()
```

```python
reg.score(df[['Weight']],df[['Price']])
```
